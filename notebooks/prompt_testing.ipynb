## Prompt Effectiveness Analysis

```python
# Analyze prompt effectiveness
test_prompts = [
    "Write a function",  # Too vague
    "Write a Python function to calculate fibonacci numbers",  # Better
    "Write a Python function to calculate fibonacci numbers using memoization for optimization, include error handling for negative inputs",  # Most specific
]

print("Prompt Specificity Analysis:")
for i, prompt in enumerate(test_prompts, 1):
    specificity = "Low" if len(prompt.split()) < 5 else "Medium" if len(prompt.split()) < 15 else "High"
    print(f"{i}. {prompt}")
    print(f"   Specificity: {specificity}")
    print(f"   Word count: {len(prompt.split())}")
    print()
```
